```python
from pathlib import Path
from PIL import Image
from pillow_heif import register_heif_opener

# Enable HEIC/HEIF support
register_heif_opener()

apple_folder = Path(
    r"DATASET_ROOT\Apple_Banana_Dataset_(Student Generated)\Apple"
)

banana_folder = Path(
    r"DATASET_ROOT\Apple_Banana_Dataset_(Student Generated)\Banana"
)

def convert_heic_to_jpg(folder):
    converted = 0
    errors = 0

    for file in folder.iterdir():
        if file.is_file() and file.suffix.lower() == ".heic":

            try:
                # Open HEIC image
                image = Image.open(file)

                # Convert to RGB
                image = image.convert("RGB")

                # Create JPG filename
                output_file = file.with_suffix(".jpg")

                # Save JPG copy
                image.save(output_file, "JPEG", quality=95)

                converted += 1
                print(f"Converted: {file.name} -> {output_file.name}")

            except Exception as e:
                errors += 1
                print(f"ERROR: {file.name} -> {e}")

    print(f"\nConverted: {converted}")
    print(f"Errors: {errors}")

print("APPLE")
convert_heic_to_jpg(apple_folder)

print("\nBANANA")
convert_heic_to_jpg(banana_folder)
```

```python
from PIL import Image
from pathlib import Path

def quality_check(folder, label):
    total = 0
    valid = 0
    corrupted = []
    small_images = []

    for file in folder.iterdir():

        if not file.is_file():
            continue

        if file.suffix.lower() not in [".jpg", ".jpeg"]:
            continue

        total += 1

        try:
            with Image.open(file) as img:

                # Verify that the image can be read
                img.verify()

            # Reopen after verify()
            with Image.open(file) as img:
                width, height = img.size

                valid += 1

                # Flag unusually small images
                if width < 100 or height < 100:
                    small_images.append(
                        (file.name, width, height)
                    )

        except Exception as e:
            corrupted.append(
                (file.name, str(e))
            )

    print(f"\n===== {label} =====")
    print("Total images:", total)
    print("Valid images:", valid)
    print("Corrupted images:", len(corrupted))
    print("Very small images:", len(small_images))

    if corrupted:
        print("\nCorrupted files:")
        for item in corrupted:
            print(item)

    if small_images:
        print("\nVery small images:")
        for item in small_images:
            print(item)


quality_check(apple_folder, "APPLE")
quality_check(banana_folder, "BANANA")
```

```python
import hashlib
from pathlib import Path
from collections import defaultdict

def calculate_hash(file):
    hash_md5 = hashlib.md5()

    with open(file, "rb") as f:
        for chunk in iter(lambda: f.read(8192), b""):
            hash_md5.update(chunk)

    return hash_md5.hexdigest()


def find_duplicates(folders):
    hashes = defaultdict(list)

    for label, folder in folders.items():

        for file in folder.iterdir():

            if file.is_file() and file.suffix.lower() in [".jpg", ".jpeg"]:

                file_hash = calculate_hash(file)

                hashes[file_hash].append(
                    (label, file.name)
                )

    duplicates = {
        h: files
        for h, files in hashes.items()
        if len(files) > 1
    }

    return duplicates


folders = {
    "Apple": apple_folder,
    "Banana": banana_folder
}

duplicates = find_duplicates(folders)

print("Number of duplicate groups:", len(duplicates))

for i, (file_hash, files) in enumerate(duplicates.items(), start=1):

    print(f"\nDuplicate group {i}:")

    for label, filename in files:
        print(f"  {label}: {filename}")
```

```python
import pandas as pd
import shutil
from pathlib import Path
from sklearn.model_selection import train_test_split

# ============================================================
# 1. PATHS
# ============================================================

combined_file = Path(
    r"DATASET_ROOT\Metadata\combined_metadata.csv"
)

student_root = Path(
    r"DATASET_ROOT\Apple_Banana_Dataset_(Student Generated)"
)

public_root = Path(
    r"DATASET_ROOT\Apple_Banana_Dataset_(Public)"
)

final_root = Path(
    r"DATASET_ROOT\Apple_Banana_Dataset_(Final)"
)

# ============================================================
# 2. LOAD METADATA
# ============================================================

df = pd.read_csv(combined_file)

print("Total images in metadata:", len(df))

# ============================================================
# 3. SELECT BALANCED DATA
# ============================================================

selected_parts = []

# 638 is the maximum balanced number available
# for BOTH Apple and Banana in BOTH sources.
target_per_class = 638

for source in ["Student", "Public"]:
    for label in ["Apple", "Banana"]:

        subset = df[
            (df["source_type"] == source) &
            (df["label"] == label)
        ].copy()

        print(f"{source} {label}: {len(subset)} available")

        selected = subset.sample(
            n=target_per_class,
            random_state=42
        )

        selected_parts.append(selected)

selected_df = pd.concat(
    selected_parts,
    ignore_index=True
)

# ============================================================
# 4. CHECK BALANCE
# ============================================================

print("\nSelected dataset:")
print("Total:", len(selected_df))

print("\nBy source:")
print(selected_df["source_type"].value_counts())

print("\nBy class:")
print(selected_df["label"].value_counts())

print("\nSource + class:")
print(
    pd.crosstab(
        selected_df["source_type"],
        selected_df["label"]
    )
)

# ============================================================
# 5. CREATE 70/15/15 SPLIT
# ============================================================

train_df, temp_df = train_test_split(
    selected_df,
    test_size=0.30,
    random_state=42,
    stratify=selected_df[["source_type", "label"]]
)

validation_df, test_df = train_test_split(
    temp_df,
    test_size=0.50,
    random_state=42,
    stratify=temp_df[["source_type", "label"]]
)

train_df = train_df.copy()
validation_df = validation_df.copy()
test_df = test_df.copy()

train_df["split"] = "train"
validation_df["split"] = "validation"
test_df["split"] = "test"

final_metadata = pd.concat(
    [train_df, validation_df, test_df],
    ignore_index=True
)

# ============================================================
# 6. CREATE FINAL FOLDERS
# ============================================================

for split in ["train", "validation", "test"]:
    for label in ["Apple", "Banana"]:

        folder = final_root / split / label
        folder.mkdir(parents=True, exist_ok=True)

# ============================================================
# 7. COPY IMAGES
# ============================================================

def get_image_path(row):

    if row["source_type"] == "Student":
        root = student_root
    else:
        root = public_root

    return root / row["label"] / row["filename"]


missing_files = 0

for _, row in final_metadata.iterrows():

    source_file = get_image_path(row)

    destination_folder = (
        final_root /
        row["split"] /
        row["label"]
    )

    destination_file = destination_folder / row["filename"]

    if not source_file.exists():
        print("WARNING - file not found:", source_file)
        missing_files += 1
        continue

    shutil.copy2(
        source_file,
        destination_file
    )

print("\nImages copied successfully!")
print("Missing files:", missing_files)

# ============================================================
# 8. SAVE FINAL METADATA
# ============================================================

metadata_folder = final_root / "Metadata"
metadata_folder.mkdir(exist_ok=True)

final_metadata_file = metadata_folder / "final_metadata.csv"

final_metadata.to_csv(
    final_metadata_file,
    index=False
)

print("\nFinal metadata saved:")
print(final_metadata_file)

# ============================================================
# 9. FINAL COUNTS
# ============================================================

print("\nFinal split counts:")
print(final_metadata["split"].value_counts())

print("\nSplit + class:")
print(
    pd.crosstab(
        final_metadata["split"],
        final_metadata["label"]
    )
)

print("\nSplit + source:")
print(
    pd.crosstab(
        final_metadata["split"],
        final_metadata["source_type"]
    )
)
```

```python
# ============================================================
# FINAL DATASET: BALANCE + SPLIT + COPY + METADATA
# ============================================================

import os
import shutil
import pandas as pd
from sklearn.model_selection import train_test_split

# ------------------------------------------------------------
# 1. PATHS
# ------------------------------------------------------------

base_path = r"DATASET_ROOT"

metadata_path = os.path.join(
    base_path,
    "Metadata",
    "combined_metadata.csv"
)

student_root = os.path.join(
    base_path,
    "Apple_Banana_Dataset_(Student Generated)"
)

public_root = os.path.join(
    base_path,
    "Apple_Banana_Dataset_(Public)"
)

final_root = os.path.join(
    base_path,
    "Apple_Banana_Dataset_(Final)"
)

# ------------------------------------------------------------
# 2. LOAD COMBINED METADATA
# ------------------------------------------------------------

df = pd.read_csv(metadata_path)

print("Total images in metadata:", len(df))
print()

# ------------------------------------------------------------
# 3. TARGET NUMBER
# ------------------------------------------------------------
# Public Banana has only 636 images.
# Therefore 636 is the maximum perfectly balanced number.

target_per_class = 636

print("Target per source/class:", target_per_class)
print()

# ------------------------------------------------------------
# 4. CHECK AVAILABLE IMAGES
# ------------------------------------------------------------

for source in ["Student", "Public"]:
    for label in ["Apple", "Banana"]:

        subset = df[
            (df["source_type"] == source) &
            (df["label"] == label)
        ]

        print(
            f"{source} {label}: "
            f"{len(subset)} available"
        )

        if len(subset) < target_per_class:
            raise ValueError(
                f"Not enough images for {source} {label}. "
                f"Only {len(subset)} available."
            )

print()

# ------------------------------------------------------------
# 5. SELECT 636 FROM EACH SOURCE + CLASS
# ------------------------------------------------------------

selected_parts = []

for source in ["Student", "Public"]:
    for label in ["Apple", "Banana"]:

        subset = df[
            (df["source_type"] == source) &
            (df["label"] == label)
        ].copy()

        selected = subset.sample(
            n=target_per_class,
            random_state=42
        )

        selected_parts.append(selected)

        print(
            f"Selected {source} {label}: "
            f"{len(selected)}"
        )

selected_df = pd.concat(
    selected_parts,
    ignore_index=True
)

print()
print("Total selected images:", len(selected_df))
print()

# ------------------------------------------------------------
# 6. CREATE SOURCE + CLASS STRATIFICATION COLUMN
# ------------------------------------------------------------

selected_df["stratum"] = (
    selected_df["source_type"] + "_" +
    selected_df["label"]
)

# ------------------------------------------------------------
# 7. FIRST SPLIT: 70% TRAIN / 30% TEMP
# ------------------------------------------------------------

train_df, temp_df = train_test_split(
    selected_df,
    test_size=0.30,
    random_state=42,
    stratify=selected_df["stratum"]
)

# ------------------------------------------------------------
# 8. SECOND SPLIT: TEMP INTO VALIDATION + TEST
# ------------------------------------------------------------

validation_df, test_df = train_test_split(
    temp_df,
    test_size=0.50,
    random_state=42,
    stratify=temp_df["stratum"]
)

# Remove temporary column
train_df = train_df.drop(columns=["stratum"])
validation_df = validation_df.drop(columns=["stratum"])
test_df = test_df.drop(columns=["stratum"])

print("Split sizes:")
print("Train:", len(train_df))
print("Validation:", len(validation_df))
print("Test:", len(test_df))
print()

# ------------------------------------------------------------
# 9. CREATE FINAL FOLDER STRUCTURE
# ------------------------------------------------------------

for split in ["train", "validation", "test"]:
    for label in ["Apple", "Banana"]:

        folder = os.path.join(
            final_root,
            split,
            label
        )

        os.makedirs(folder, exist_ok=True)

# Metadata folder
metadata_final = os.path.join(
    final_root,
    "Metadata"
)

os.makedirs(metadata_final, exist_ok=True)

# ------------------------------------------------------------
# 10. FUNCTION TO FIND ORIGINAL IMAGE
# ------------------------------------------------------------

def find_image(row):

    source = row["source_type"]
    label = row["label"]
    filename = row["filename"]

    if source == "Student":
        root = student_root
    else:
        root = public_root

    image_path = os.path.join(
        root,
        label,
        filename
    )

    return image_path


# ------------------------------------------------------------
# 11. COPY IMAGES INTO FINAL DATASET
# ------------------------------------------------------------

def copy_images(split_df, split_name):

    copied = 0
    missing = 0

    for _, row in split_df.iterrows():

        source_file = find_image(row)

        # Create a unique filename so images from
        # different sources cannot overwrite each other.

        source_prefix = row["source_type"].upper()

        new_filename = (
            source_prefix + "_" +
            str(row["image_id"]) + "_" +
            row["filename"]
        )

        destination_folder = os.path.join(
            final_root,
            split_name,
            row["label"]
        )

        destination_file = os.path.join(
            destination_folder,
            new_filename
        )

        if os.path.exists(source_file):

            shutil.copy2(
                source_file,
                destination_file
            )

            copied += 1

        else:

            print("MISSING:", source_file)
            missing += 1

    print(
        f"{split_name}: "
        f"{copied} copied, "
        f"{missing} missing"
    )


# Copy all three splits

copy_images(train_df, "train")
copy_images(validation_df, "validation")
copy_images(test_df, "test")

# ------------------------------------------------------------
# 12. ADD SPLIT INFORMATION TO METADATA
# ------------------------------------------------------------

train_df["split"] = "train"
validation_df["split"] = "validation"
test_df["split"] = "test"

final_metadata = pd.concat(
    [
        train_df,
        validation_df,
        test_df
    ],
    ignore_index=True
)

# ------------------------------------------------------------
# 13. SAVE FINAL METADATA
# ------------------------------------------------------------

final_metadata_path = os.path.join(
    metadata_final,
    "final_metadata.csv"
)

final_metadata.to_csv(
    final_metadata_path,
    index=False
)

# ------------------------------------------------------------
# 14. PRINT FINAL RESULTS
# ------------------------------------------------------------

print()
print("=" * 60)
print("FINAL DATASET CREATED")
print("=" * 60)

print()
print("Total images:", len(final_metadata))

print()
print("By split:")
print(
    final_metadata["split"]
    .value_counts()
    .sort_index()
)

print()
print("By class:")
print(
    final_metadata["label"]
    .value_counts()
)

print()
print("By source:")
print(
    final_metadata["source_type"]
    .value_counts()
)

print()
print("Source + class:")
print(
    pd.crosstab(
        final_metadata["source_type"],
        final_metadata["label"]
    )
)

print()
print("Split + class:")
print(
    pd.crosstab(
        final_metadata["split"],
        final_metadata["label"]
    )
)

print()
print("Split + source:")
print(
    pd.crosstab(
        final_metadata["split"],
        final_metadata["source_type"]
    )
)

print()
print("Final metadata saved to:")
print(final_metadata_path)

print()
print("FINAL DATASET LOCATION:")
print(final_root)

print()
print("DONE!")
```

```python
import tensorflow as tf
from tensorflow.keras import layers

# --------------------------------------------------
# 1. Final dataset location
# --------------------------------------------------

final_root = r"DATASET_ROOT\Apple_Banana_Dataset_(Final)"

train_path = final_root + r"\train"
validation_path = final_root + r"\validation"
test_path = final_root + r"\test"


# --------------------------------------------------
# 2. Settings
# --------------------------------------------------

IMG_SIZE = (128, 128)
BATCH_SIZE = 32
SEED = 42


# --------------------------------------------------
# 3. Load training dataset
# --------------------------------------------------

train_ds = tf.keras.utils.image_dataset_from_directory(
    train_path,
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    label_mode="binary",
    shuffle=True,
    seed=SEED
)


# --------------------------------------------------
# 4. Load validation dataset
# --------------------------------------------------

validation_ds = tf.keras.utils.image_dataset_from_directory(
    validation_path,
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    label_mode="binary",
    shuffle=False
)


# --------------------------------------------------
# 5. Load test dataset
# --------------------------------------------------

test_ds = tf.keras.utils.image_dataset_from_directory(
    test_path,
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    label_mode="binary",
    shuffle=False
)


# --------------------------------------------------
# 6. Display class names
# --------------------------------------------------

print("Class names:")
print(train_ds.class_names)


# --------------------------------------------------
# 7. Normalise pixel values
# --------------------------------------------------

normalization_layer = layers.Rescaling(1./255)

train_ds = train_ds.map(
    lambda x, y: (normalization_layer(x), y)
)

validation_ds = validation_ds.map(
    lambda x, y: (normalization_layer(x), y)
)

test_ds = test_ds.map(
    lambda x, y: (normalization_layer(x), y)
)


# --------------------------------------------------
# 8. Improve data pipeline performance
# --------------------------------------------------

AUTOTUNE = tf.data.AUTOTUNE

train_ds = train_ds.prefetch(AUTOTUNE)
validation_ds = validation_ds.prefetch(AUTOTUNE)
test_ds = test_ds.prefetch(AUTOTUNE)


# --------------------------------------------------
# 9. Final confirmation
# --------------------------------------------------

print()
print("Preprocessing pipeline created successfully!")
print("Image size:", IMG_SIZE)
print("Batch size:", BATCH_SIZE)
print("Pixel range: 0-1")
print("Classes:", ["Apple", "Banana"])
```
